```python
import PyPDF2

def extract_text_from_pdf(pdf_path):
    # Create a PDF file reader object
    with open(pdf_path, 'rb') as file:
        pdf_reader = PyPDF2.PdfReader(file)
        text = ""

        # Iterate through all the pages and extract text
        for page_num in range(len(pdf_reader.pages)):
            page = pdf_reader.pages[page_num]
            text += page.extract_text()

    return text

# Path to your PDF file
pdf_path = "/content/NTS Experiecnce Letter.pdf"

# Extract text content
text_content = extract_text_from_pdf(pdf_path)
print(text_content)
```

```python
!pip install sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode([text_content])
embeddings
```

```python
import uuid
document_id = str(uuid.uuid4())
```

```python
import chromadb
from chromadb.config import Settings

# Initialize the Chroma client
client = chromadb.Client(Settings())

# Create a new collection or get an existing one
collection = client.create_collection("pdf_embeddings_new")

# Add the document and its embeddings to the collection
collection.add(documents=[text_content], embeddings=embeddings.tolist(), ids=[document_id])

print('Embeddings are successfully stored in Chroma DB')
```

```python
from flask import Flask, request, jsonify
from sentence_transformers import SentenceTransformer
import chromadb
from chromadb.config import Settings

app = Flask(__name__)
model = SentenceTransformer('all-MiniLM-L6-v2')

# Initialize the Chroma client
client = chromadb.Client(Settings())
collection = client.get_collection("pdf_embeddings_new")

@app.route('/query', methods=['POST'])
def query_embeddings():
    data = request.get_json()
    question = data['question']
    question_embedding = model.encode([question])
    results = collection.query(embedding=question_embedding.tolist(), top_k=1)
    return jsonify(results)

if __name__ == '__main__':
    app.run(debug=True)
```
